## Control Flow

### IF/ELSE/ELIF

if, elif, and else help Python make decisions.

- if – Executes a block of code when a condition is true.
- elif – Checks another condition if the previous if or elif condition is false.
- else – Executes a block of code when none of the conditions are true.

Real-time analogy:

- If it is raining, take an umbrella.

- Else if it is sunny, wear sunglasses.

- Else – carry on normally.

Python checks each condition one by one. As soon as one condition is true, it performs that action.
##### Example:

```python
age = int(input("Enter Your Age:"))

if age < 18:
    print("Minor")
elif age == 18:
    print("Just became an Adult")
else:
    print("Adult")
```

### While Loop
A **while loop** is used to repeatedly execute a block of code as long as a given condition is `true`.

real-time analogy:
A while loop is like continuing to add water to a tank while the tank is not full and stopping once the tank becomes full.


```python
secrete_number = 50

while True:
    user_input = int(input("Enter Your Guess:"))
    if secrete_number == user_input:
        print("Yes, Your Guess is Right")
        break
    else:
        print("Sorry, Your Guess is Wrong")
```

This code is a simple number guessing game.

- Stores the secret number 62 in the variable `secret_number`.

- `while True` - So, the loop will continue running indefinitely until we explicitly stop it using break.

- `input()` asks the user to enter a number.
`int()` converts the user's input from a string into an integer.
The entered number is stored in `user_input`.

- Checks whether the user's guess is equal to the secret number.
If the guess is 62, this message is displayed.
Stops the while loop because the user guessed correctly.
If the guess is not equal to 62, the else block runs.
Displays a message saying the guess is incorrect. Then the loop starts again and asks for another guess.

The program keeps asking the user to guess the secret number until the user enters 62. Once the correct number is entered, break stops the loop.

### Break

The `break` statement is used when we want to stop a loop immediately, even if the loop could continue running.

Once Python reaches `break`, it completely exits the current loop and continues with the code written after the loop.

##### Real-world analogy

Imagine you are searching for a particular book in a library.

You check the books one by one:

- Book 1 → Not found
- Book 2 → Not found
- Book 3 → Found!
- Book 4 → No need to check
- Book 5 → No need to check

Once you find the book, you stop searching.

That's exactly what `break` does.

##### Example

```python
# Immediately exit the loop if condition is true
for number in range(1, 11):
    if number == 6:
        break

    print(number)

print("Loop has ended")
```

### Continue

The `continue` statement is used when we want to skip the current iteration of a loop and move to the next iteration.

`continue` does not stop the entire loop. It only skips the current round.

##### Real-world analogy

Imagine a teacher checking attendance.

The teacher goes through every student:

- Student 1 → Present
- Student 2 → Present
- Student 3 → Absent → Skip
- Student 4 → Present
- Student 5 → Present

The teacher doesn't stop taking attendance just because one student is absent.

##### Example

```python
# Skip the Condition 
for number in range(1, 11):
    if number == 5:
        continue

    print(number)
```

### Pass 

`Pass` basically means **Do nothing for now**.

It is used when Python expects some code inside a block, but we don't want to write any action yet.

It doesn't stop the loop, and it doesn't skip to the next iteration.

It simply does nothing and allows Python to continue normally.

##### Real-world analogy

Imagine you are making a checklist for a project.

You have:

- Login page → Completed
- Home page → Completed
- Payment page → Work in progress
- Profile page → Completed

You know that you need to work on the payment page later, but you haven't decided what code to write yet.

You can temporarily use pass as a placeholder.

##### Example

```python
# Placeholder
for number in range(1, 6):

    if number == 3:
        pass

    print(number)
```

### List Comprehension

List comprehension is a shorter and cleaner way of creating a list using a loop.

Normally, if we want to create a list of squares:

```python
squares = [] 
for number in range(1, 6): 
    squares.append(number * number)
print(squares)
```

We can write the same idea using list comprehension:

```python
squares_list = [number * number for number in range(1, 6)]
print(squares_list)
```

### Real-World Analogy

Imagine you have a box containing numbers:

`1  2  3  4  5`

You want another box containing their squares:

`1  4  9  16  25`

You are essentially saying:

"Take every number and perform this operation on it."

#### Basic Syntax
`[expression for item in iterable]`

##### Example

```python
numbers = [1, 2, 3, 4, 5] 
two_table = [number * 2 for number in numbers] 
print(two_table)
```

#### List Comprehension with a Condition



```python
#List comprehension with if condition
even_numbers = [
    number
    for number in range(1, 21)
    if number % 2 == 0
]
print(even_numbers)
```

```python
#List Comprehension with if-else

Odd_Even_num = [
    "Even" if number % 2 == 0 else "Odd"
    for number in range(1,11)
]

print(Odd_Even_num)
```

```python
#String Example
names = ["Ajay", "priya", "arun"]

uppercase_names = [name.upper() for name in names]

print(uppercase_names)
```

### Dictionary Comprehension

Dictionary comprehension is similar to list comprehension, but instead of creating a list, it creates a dictionary.

It allows us to create key-value pairs using a compact syntax.

#### Basic Syntax
{key: value for item in iterable}

##### Example

```python
squares = { number: number * number for number in numbers } 
print(squares)
```

### Dictionary Comprehension with a Condition

```python
# Dictionary Comprehension wiht If Condition

numbers = range(1, 11) 
even_squares = { number: number * number for number in numbers if number % 2 == 0 } 
print(even_squares)
```

```python
#Dictionary Comprehension with Strings
names = ["Ajay", "Priya", "Arun"]

name_lengths = {
    name: len(name)
    for name in names
}

print(name_lengths)
```

### Python Functions

A function is one of the most useful concepts in Python.

When we write a program, we often need to perform the same task multiple times. Instead of writing the same code again and again, we can put that code inside a function and reuse it whenever we need it.

For example, if we need to calculate a student's total marks several times, we can create a function for calculating the total.

```python
def calculate_total(mark1, mark2):
    return mark1 + mark2

print(calculate_total(80, 90)) 
print(calculate_total(70, 85))
```

This makes our program more organized, reusable, and easier to maintain.

### Arguments

Sometimes a function needs information from outside to perform its task.

That information is passed into the function using arguments.

For example, instead of creating a function that always adds 10 + 20, we can tell the function which two numbers to add.

```python
def add_numbers(a, b):
    print("Sum of a + b:",a + b)
add_numbers(10, 20) 
add_numbers(50, 30)
```

add_numbers(10, 20) ---> `10` and `20` are arguments
add_numbers(50, 30) ---> `50` and `30` are arguments

#### Parameter vs Argument

These two terms are often confused.

##### Parameter

A `parameter` is the variable written when defining the function.

def greet(`name`):

    print("Hello", name)

Here, `name` is a parameter.

##### Argument

An `argument` is the actual value passed when calling the function.

greet(`"Rahul"`)

Here, `"Rahul"` is an argument.

Easy way to remember:

Function definition  → `Parameter`

Function call        → `Argument`

### Real-World Analogy

Imagine ordering food.

What pizza do you want?

What size?

You provide:

Pizza = Margherita

Size = Large

The questions are like `parameters`.

The values you provide are like `arguments`.

### Default Arguments

A default argument is a value that a function uses automatically when we don't provide a value for that parameter.


```python

def greet(name="Guest"):
    print("Hello", name)
#If we provide a name:
greet("Akilesh")
#If we don't provide a name:
greet()

#The function uses "Guest" because it was given as the default value.
```

#### Multiple Default Arguments

```python
def student_details(name="Unknown", course="Python"): 
    print("Name:", name) 
    print("Course:", course)
#If we don't provide a name:    
print("Dont provide a value:")
student_details()

#If we provide a name:
print("\nProvide a value:")
student_details("Akilesh", "Data Science")
```

### Recursive Functions

A `recursive function` is a function that calls itself.
it is useful for problems that can be broken into smaller versions of the same problem.

A recursive function normally needs two important parts:

- `Base case` – tells the function when to stop.
- `Recursive case` – calls the function again with a smaller or simpler problem.

Without a base case, the function can keep calling itself indefinitely.

### Real-World Analogy

Imagine standing between two mirrors.

You can see your reflection inside another reflection, and so on.

But in programming, we need to tell the process when to stop.

The base case acts like the stopping point.

##### Recursive Example

```python
def factorial(number):

    if number == 1:
        return 1

    return number * factorial(number - 1)

result = factorial(5)

print("Factorial:", result)
```

### Iterators

An `iterator` is an object that allows us to access items one at a time.

Instead of getting all the values at once, an iterator remembers where it currently is and gives us the next value whenever we ask for it.

Python uses two important functions with iterators:

#### Real World Example

### Generators

A `generator` is a simple way of creating an iterator.

Instead of creating and storing all values in memory at once, a generator produces values one at a time when needed.

This makes generators especially useful when working with:

- Large amounts of data
- Large files
- Data streams
- Infinite sequences

**Key point:** Generator = Produce values when needed.

```python
def generate_numbers(): 
    for num in range(1,11):
        yield num * num
gen_num = generate_numbers()

for gen in gen_num:
    print(gen)
```

### yield

`yield` is used inside a generator function.

It is similar to return, but there is an important difference.

`return`:

Gives the result and completely ends the function.


`yield`:

Gives a value temporarily and pauses the function.

When the next value is requested, the function continues from where it stopped.

**key Point:** yield = Give now, continue later.

##### Example

```python
def show_numbers(): 
    print("Starting") 
    yield 10 
    print("Continuing") 
    yield 20 
    print("Finished") 
    yield 30 
    
numbers = show_numbers() 
print(next(numbers))
print(next(numbers))
print(next(numbers))
```

### Decorators

A `decorator` is a function that adds extra behavior to another function without changing the original function's code.

### Real-world analogy

Imagine you have a basic coffee:

Coffee

You can add:

Coffee + Milk

Coffee + Sugar

Coffee + Whipped Cream

The original coffee hasn't changed.

**Key point:** Decorator = Add extra behavior without modifying the original function.

```python
def uppercase(function):
    def innerfunc():
        print("Hello")
        return function().upper()
    return innerfunc

@uppercase
def name():
    return "Akilesh"

greet = name()
print(greet)
```

#### Decorator with arguments

```python
def log(func):
    def wrapper(*args,**kwargs):
        print("function started!")

        result = func(*args,**kwargs)

        print("Function ended")

        return result
    return wrapper

@log
def add(a,b):
    return a+b

sum = add(b=30,a=40)

print(sum)
```

### Closures

A closure happens when an inner function remembers values from its outer function, even after the outer function has finished executing.

**Key point:** Closure = Inner function remembers outer function's data.

```python
def create_multiplier(number): 
    def multiply(value): 
        return value * number 
    return multiply 
double = create_multiplier(2)
print(double(5))
print(double(100))
```

### global and nonlocal

These keywords are related to variable scope.

Global Scope -> Function Scope -> Nested Function Scope

`global`

A variable created outside a function normally belongs to the global scope.

```python
count = 0 #Global Scope Variable
def increase_count(): 
    global count # If we need to call the global scope variable inside a function we need to use global keyword.
    count += 1 
increase_count() 
increase_count() 

print(count)
```

### nonlocal

`nonlocal` is used inside a nested function when we want to modify a variable belonging to the outer function.



```python
def counter(): 
    count = 0 #Local Scope Variable
    def increase(): 
        nonlocal count #If we need to reassign the outer scope varibale we need to use nonlocal keyword.
        count += 1 
        return count 
    return increase 
my_counter = counter() 
print(my_counter()) 
print(my_counter()) 
print(my_counter())
```

**Key point:** 

global = outside all functions

nonlocal = outside the current nested function

**`filter()`**

`filter()` is used when we want to select only the items that satisfy a condition.

Collection --> Check condition --> Keep matching values.

True --> keep the items

`filter(function, iterable)`

#### Real World analogy

```python
marks = [45, 78, 32, 90, 65] 
def is_pass(mark): 
    return mark >= 50 
passed_marks = filter(is_pass, marks) 
print(list(passed_marks))
```
