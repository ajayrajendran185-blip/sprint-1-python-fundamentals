# Module 4 - Logging

## What is Logging?

### Logging is a way to track events that happens during the ececution of a program.

### Example

#### imagine the security guard maintaing a register at a company entrance.

#### security gauard records:

#### Time - when someone entered
#### person - who entered
#### Activity - What they did
#### problem - Any unusual activity

```python
import logging

logging.basicConfig(filename="security.log",level=logging.INFO,)

logging.info("Ajay entered the company")
logging.info("Ajay went to the meeting room")
logging.info("Ajay completed his meeting")
logging.warning("Unknown person tried to enter")
logging.info("Ajay left the company")

print("Security activities recorded successfully!")
```

# why use Logging ?

### Logging Used to:

##### Find and fix eroors
##### Keep a record of program activities
##### Monitor what the program is doing
##### Help in Debugging problems



# Print() vs Logging

### print is mainly for displaying a messaages while logging is used to properly track and records what happened in programs.

### Example:

### Imagine a shop

#### Print() = Shop Announcement
#### The shopkeeper Says Shop Open at 9:00AM and Close at 9:00PM

#### Logging = Shop Register
#### The shopkeeper records
#### =>Shop opened at 9:00AM
#### =>Product Sold
#### =>Payment Failed
#### =>Shop Closed at 9:00PM


```python
import logging

logging.basicConfig(filename="shop.log",level=logging.info)

print("shop opens at 9:00AM")

logging.info("shop opened at 9:00am")
logging.info("customer entered the shop")
logging.info("product sold")
logging.warning("payment failed")
logging.info("shop closed at 9:00PM")

print("shop close at 9:00PM")
```

# Log Levels

## DEBUG
### Detailed information used to find and fix problems in program.

## Example
### A security guard Writes:
### =>Checked ID card 
### =>ID is valid
### =>Checked entry Time
### =>Door opened

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Checking username")
logging.debug("Connecting to database")
logging.info("User logged in")
```

## INFO
###  It is use to record normal and important events happening in a program.

## Example 

## A security guard notes

### => Arun entered the company
### => Arun went for lunch
### => Arun left the Company

```python
import logging

logging.basicConfig(filename="security.log",level=logging.INFO,)

logging.info("Arun entered the company")
logging.info("Arun went for lunch")
logging.info("Arun left the company")

print("Activities recorded successfully!")
```
