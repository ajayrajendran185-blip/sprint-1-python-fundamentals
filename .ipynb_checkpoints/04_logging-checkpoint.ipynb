# Module 4 - Logging

## What is Logging?

### Logging is a way to track events that happens during the ececution of a program.

### Example

#### imagine the security guard maintaing a register at a company entrance.

#### security gauard records:

#### Time - when someone entered
#### person - who entered
#### Activity - What they did
#### problem - Any unusual activity

```python
import logging

logging.basicConfig(filename="security.log",level=logging.INFO,)

logging.info("Ajay entered the company")
logging.info("Ajay went to the meeting room")
logging.info("Ajay completed his meeting")
logging.info("Ajay left the company")

print("Security activities recorded successfully!")
```

# why use Logging ?

### Logging Used to:

##### Find and fix eroors
##### Keep a record of program activities
##### Monitor what the program is doing
##### Help in Debugging problems



# Print() vs Logging

### print is mainly for displaying a messaages while logging is used to properly track and records what happened in programs.

### Example:

### Imagine a shop

#### Print() = Shop Announcement
#### The shopkeeper Says Shop Open at 9:00AM and Close at 9:00PM

#### Logging = Shop Register
#### The shopkeeper records
#### =>Shop opened at 9:00AM
#### =>Product Sold
#### =>Payment Failed
#### =>Shop Closed at 9:00PM


```python
import logging

logging.basicConfig(filename="shop.log",level=logging.info)

print("shop opens at 9:00AM")

logging.info("shop opened at 9:00am")
logging.info("customer entered the shop")
logging.info("product sold")
logging.warning("payment failed")
logging.info("shop closed at 9:00PM")

print("shop close at 9:00PM")
```

# Log Levels

## DEBUG
### Detailed information used to find and fix problems in program.

## Example
### A security guard Writes:
### =>Checked ID card 
### =>ID is valid
### =>Checked entry Time
### =>Door opened

```python
import logging

logging.basicConfig(level=logging.DEBUG)

logging.debug("Checking username")
logging.debug("Connecting to database")
logging.info("User logged in")
```

## INFO
###  It is use to record normal and important events happening in a program.

## Example 

## A security guard notes

### => Arun entered the company
### => Arun went for lunch
### => Arun left the Company

```python
import logging

logging.basicConfig(filename="security.log",level=logging.INFO,)

logging.info("Arun entered the company")
logging.info("Arun went for lunch")
logging.info("Arun left the company")

print("Activities recorded successfully!")
```

# Warning

### It is used to watch something unusual happened that may cause a problem, so we should pay attention

# Example

### The Security Guard Notices:
#### => An unknown person is standing near the entrance
#### It doesn't mean somthing has gone wrong yet, but the security should pay attention.

```python
import logging

logging.basicConfig(filename="security.log",level=logging.INFO)

logging.info("Ajay entered the company")

logging.warning("Unknown person is standing near the entrance")

logging.info("Security guard checked the visitor")
```

# Error

## It means something went wrong in the program and needs attention or fixing.

# Example 

### A security tries to open the gate but the gate lock is broken
### => " Main Gate Cannot be opened"

### This is an error because something went wrong and needs to be fixed

```python
import logging

logging.basicConfig(filename="security.log",level=logging.INFO )

logging.info("Security guard started duty")

logging.error("Main gate lock is broken")

logging.info("Security guard contacted maintenance team")
```

# Critical

## It means Very Serious Problem That Requires Immediate Attention.

# Example 

### A security Guard Notices:

### => " fire detected inside the building "

### This is more serious than a normal warning or error. Immediate action required

```python
import logging

logging.basicConfig(filename="security.log",level=logging.INFO,)

logging.info("Security guard started duty")

logging.critical("Fire detected inside the building")

logging.info("Emergency team has been contacted")
```

# Log Formatters

## A log formatter defines how a log message should look when it is displayed or saved

# Example

## Instead Of Getting
### => user Logged In

## You Can format it as
### => 2026-09-10 10:40:15 - INFO - user logged in

```python
import logging

logging.basicConfig(level=logging.INFO,format="%(asctime)s - %(levelname)s - %(message)s")

logging.info("User logged in")
```

# File Handlers

## File handler is used when you want to store logs in a file instead of showing them only on screen.

# Example

### Imagine you are working in a bank 

### Every time a customer performs a transaction , the bank wants to keep a record in a register for future references. 

### For Example
#### customer deposited Rs.50000
#### customer withdraw  Rs.20000
#### Transaction failed

```python
import logging

logging.basicConfig(filename="bank.log",level=logging.INFO)

logging.info("Money deposited")

logging.info("Money withdrawn")

logging.error("Transaction failed")
```

# Console Handler

### A console handler is used to display log messages directly on the screen while your python programming  is running.

# Example

### Imagine the security watching CCTV

### When something happens unusual,the guard sees it immediately on the monitor. 

```python
import logging

logger = logging.getLogger()

console_handler = logging.StreamHandler()

logger.addHandler(console_handler)

logger.setLevel(logging.INFO)

logger.info("User logged in")
logger.warning("Low battery")
logger.error("Payment failed")
```
