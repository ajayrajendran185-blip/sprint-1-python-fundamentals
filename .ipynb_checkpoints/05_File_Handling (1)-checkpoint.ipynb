# Notebook 5 – File Handling

## 1. Introduction to File Handling

### What is File Handling?

`File handling` means creating, reading, writing, updating, and managing files using Python.

Files allow us to store data permanently so that it remains available even after the Python program stops.

**Common file types**

- .txt   → Text files
- .csv   → Tabular data
- .json  → Structured data
- .pkl   → Python objects

**Why do we need File Handling?**

For example, an AI/ML application may need to:

- Read a dataset
- Save processed data
- Load configuration files
- Store model objects
- Create log files
- Manage folders and files

There are four different methods (modes) for opening a file:

"r" - Read - Default value. Opens a file for reading, error if the file does not exist

"a" - Append - Opens a file for appending, creates the file if it does not exist

"w" - Write - Opens a file for writing, creates the file if it does not exist

"x" - Create - Creates the specified file, returns an error if the file exists

### Reading Text Files

Python provides the `open()` function for working with files.

By default the `read()` method returns the whole text, but you can also specify how many characters you want to return:

`read(5)`


`Methods:`

- read()      - Reads entire file
- readline()  - Reads one line
- readlines() - Reads all lines as a list

```python
file = open("student.txt", "r")

content = file.read()

print(content)

file.close()
```

### Writing Text Files

Use "w" mode to write to a file.

**Modes:**

- w - Write (overwrite)
- a - Append(add at the end)
- x - Create new file

```python
file = open("students.txt", "w")

file.write("Akilesh\n")
file.write("Rahul\n")
file.write("Priya\n")

file.close()
```

### Context Manager – with

we can use a context manager:

```python
with open("students.txt", "r") as file:
    content = file.read()

print(content)
```

Python automatically closes the file after the with block.

**Why use with?**

- Automatically closes the file
- Makes code cleaner
- Helps prevent resource leaks
- Is the recommended approach

### Writing Using with

```python
with open("students.txt", "w") as file:
    file.write("Akilesh\n")
    file.write("Rahul\n")
    file.write("Priya\n")
```

### CSV Files

`CSV` stands for Comma-Separated Values.

Example:

id,name,age,course

1,Akilesh,22,Python

2,Rahul,21,Java

3,Priya,23,Python

CSV files are commonly used for:

- Datasets
- Reports
- Data analysis
- Machine learning

### Reading CSV Files

Python provides the built-in csv module.


```python
import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

### Writing CSV Files


```python
import csv

students = [
    ["id", "name", "age"],
    [1, "Akilesh", 22],
    [2, "Rahul", 21],
    [3, "Priya", 23],
    [4,"Darshini",24]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerows(students)
```

### JSON Files

JSON stands for `JavaScript Object Notation`.

JSON is commonly used for:

- APIs
- Configuration files
- Data exchange
- Storing structured information

##### JSON Example:

```python
{
    "name": "Akilesh",
    "age": 22,
    "course": "Python"
}
```

### Writing JSON

Python provides the json module.

Basic syntax is:

`json.dump(data, file)`


```python
import json

student = {
    "name": "Akilesh",
    "age": 22,
    "course": "Python"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)
```

### Reading JSON

```python
import json

with open("student.json", "r") as file:
    student = json.load(file)

print(student)
print(student["name"])
```

### Pickle Files

`pickle` is used to serialize Python objects.

Serialization means converting a Python object into a format that can be stored and loaded later.

```python
student = {
    "name": "Akilesh",
    "marks": [90, 85, 95]
}
```

We can save this Python object using pickle.

### Writing Pickle Files

```python
import pickle

student = {
    "name": "Akilesh",
    "marks": [90, 85, 95]
}

with open("student.pkl", "wb") as file:
    pickle.dump(student, file)

print("Pickle file is created")
```

### Reading Pickle Files

```python
import pickle

with open("student.pkl", "rb") as file:
    student = pickle.load(file)

print(student)
```

### pathlib

```python
#Check if File Exists

from pathlib import Path

file_path = Path("students.txt")

if file_path.exists():
    print("File exists")
else:
    print("File does not exist")
```

### shutil

shutil is used for high-level file and directory operations.

```python
import shutil
```

```python
#Copy a file
shutil.copy("students.txt", "backup_students.txt")

#Move a file
shutil.move("students.txt", "data/students.txt")

#Copy a directory
shutil.copytree("data", "data_backup")
```

```python
#Delete a directory
shutil.rmtree("data_backup")
```

### File Validation

Before reading or processing a file, we should validate it.

```python
from pathlib import Path

file_path = Path("students.csv")

if not file_path.exists():
    print("File does not exist")

elif not file_path.is_file():
    print("Path is not a file")

else:
    print("File is valid")
```

```python
#Validate File Extension
from pathlib import Path

file_path = Path("students.csv")

if file_path.suffix == ".csv":
    print("Valid CSV file")
else:
    print("Invalid file type")

# Validate File Size
from pathlib import Path

file_path = Path("students.csv")

if file_path.exists():
    if file_path.stat().st_size > 0:
        print("File contains data")
    else:
        print("File is empty")
```

This is useful when processing uploaded datasets.

## Directory Operations

We can perform several operations on directories.

```python
#Create
from pathlib import Path

Path("data").mkdir(exist_ok=True)

#Check
folder = Path("data")

if folder.is_dir():
    print("Directory exists")

#List
for item in folder.iterdir():
    print(item)

#Delete an empty directory
folder1 = Path("Empty")
folder1.rmdir()
```

## pathlib vs os

Both can work with files and directories.

| Task             | `pathlib`        | `os`                 |
| ---------------- | ---------------- | -------------------- |
| Check file       | `Path.exists()`  | `os.path.exists()`   |
| Check directory  | `Path.is_dir()`  | `os.path.isdir()`    |
| Create directory | `Path.mkdir()`   | `os.mkdir()`         |
| List files       | `Path.iterdir()` | `os.listdir()`       |
| File extension   | `Path.suffix`    | `os.path.splitext()` |
| File size        | `Path.stat()`    | `os.path.getsize()`  |
