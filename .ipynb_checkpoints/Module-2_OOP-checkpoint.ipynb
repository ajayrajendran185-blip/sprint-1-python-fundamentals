## 1. Class

### What is a Class?
A **class** is a blueprint/template used to create objects.

### Real-world analogy
think of a building plan.

an architect creates a blueprint that says:

* how many rooms?

* where are the doors?

* where are the windows?

the blueprint itself isn't a building. it is a plan for creating buildings.

similarly, a Python class is a blueprint for **creating objects.**

## concept
A class defines:

* what data an object will have
* what actions an object can perform


### Why use it?
Classes group related **data (attributes)** and **behavior (methods)** together.

### Syntax
```python
class ClassName:
    # attributes and methods
    pass
```

````python
class Employee:
    name = "Aadharsh"
    designation = "AI&ML"

print(Employee.name)
print(Employee.designation)
````

### Output
```text
Aadharsh
AI&ML
```

**Key idea:** The class describes what a Student object can contain or do.


## 2. Object

An **object** is an instance of a class.

## Real-world analogy

The architect's blueprint can be used to construct many actual buildings.

similarly, one class can create many objects.

             Student Class
                   ↓
        ┌──────────┼──────────┐
        ↓          ↓          ↓
     student1   student2   student3
      Aadharsh      vijay      gowtham
### Why use objects?
A class is the blueprint; objects are the actual entities created from that blueprint.


````python
class Employee:
    def display(self):
        print("This is a Employee object.")

Employee1 = Employee()
Employee2 = Employee()

Employee1.display()
Employee2.display()
````

### Output
```text
This is a Employee object.
This is a Employee object.
```


## 3. Constructor – `__init__()`

`__init__()` is a special method that runs automatically when an object is created.

## Real-world analogy
imagine you're creating a employee ID card.

When a new students joins, you immediately need to enter:

* Name
* Age
* Department
  
The initial information is set when the student is created.
### Why use it?
It initializes an object's starting data.


````python
class employee:
    def __init__(self, name, age):
        self.name = name
        self.age = age

employee = employee("Aadharsh", 22)

print(employee.name)
print(employee.age)
````

### Output
```text
Aadharsh
22
```

`self` refers to the **current object**.


## 4. Instance Variables

Instance variables belong to a **particular object**. Different objects can have different values.

## Real-world analogy
Consider three students:

* Aadharsh → Age 21
* vijay → Age 22
* gowtham → Age 20

Each students has their own age.

so age belongs to a particular students.

## Concept

Instance variables are variables that belong to a **specific object.**



````python
class employee:
    def __init__(self, name, phoneno):
        self.name = name
        self.phoneno = phoneno

employee1 = employee("Aadharsh", 234567)
employee2 = employee("vijay", 12345675)

print(employee1.name, employee1.phoneno)
print(employee2.name, employee2.phoneno)
````

### Output
```text
Aadharsh 234567
vijay 12345675
```

## 5. Class Variables

A class variable is shared by all objects of  class.

## Real-world analogy

Suppose all employees belong to the same company:

* Aadharsh  → stackly
* vijay  → stackly
* gowtham  → stckly

The company name is common to everyone.
 instead of storing the same information separately for every object, we can create a *class variable**
 
### When to use it?
it can be used when a value should be common to every instance.


````python
class employee:
    company = "stackly"

    def __init__(self, name):
        self.name = name

employee1 = employee("Aadharsh")
employee2 = employee("vijay")

print(employee1.name, employee1.company)
print(employee2.name, employee2.company)
````

### Output
```text
Aadharsh stackly
vijay stackly
```

## 6. Instance Methods

Instance methods operate on object-specific data and normally receive `self` as the first parameter.

## Real-world analogy

A employee can perform actions:

* prepare
* attend meeting
* completing work

  These actions belongs to the student. 

````python
class employee:

    def __init__(self, name):
        self.name = name

    def study(self):
        print(self.name, "is working.")

employee1 = employee("Aadharsh")
employee1.study()
````

### Output
```text
Aadharsh is working.
```


## 7. Static Methods

A static method does not need `self` or `cls`. It behaves like a utility function that logically belongs to the class.

Use `@staticmethod`.

## Real-world analogy

Suppose a college has a utility function:

Check whether a number is even.

This doesn’t depend on any particular student.

It doesn’t need student information.

That’s a good use case for a static method.


````python
class Calculator:
    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
````

### Output
```text
30
```


## 8. Class Methods

A class method receives the class itself as `cls`.

Use `@classmethod`.

### When to use it?
Common uses include modifying class-level data and creating alternative constructors.

## Real-world analogy

Imagine a college changes its name.

The change applies to the college as a whole, not just one student.

That’s class-level behavior.


````python
class employee:
    company = "stackly"

    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

employee.change_company("stackly pvt ltd")

print(employee.company)
````

### Output
```text
stackly pvt ltd
```


## 9. Properties – `@property`

A property allows a method to be accessed like an attribute.

It is especially useful when you want controlled access to internal data.

## Real-world analogy

Think about a bank account.

You may want to allow someone to see the balance, but you don’t want them to directly modify it however they want.

A property allows controlled access.


````python
class BankAccount:

    def __init__(self, balance):
        self._balance = balance

    @property
    def balance(self):
        return self._balance

account = BankAccount(5000)

print(account.balance)
````

### Output
```text
5000
```

## 10. Encapsulation

**Encapsulation** means bundling data and methods together and controlling how internal data is accessed.

Python uses naming conventions such as:
- `name` → public
- `_name` → protected-by-convention
- `__name` → name-mangled/private-by-convention
## Real world-analogy

Think about an **ATM.**
 You don't directly access the bank's database.
 Instead,you use:

 *  ATM
 *  ↓
*  Enter PIN
 *  ↓
*  Choose operation
 *  ↓
*  Withdraw money

the internal banking system is protected.

that's the basic idea of **encapsulation.**


````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

account = BankAccount(5000)
account.deposit(1000)

print(account.get_balance())
````

### Output
```text
6000
```

The balance is accessed through controlled methods rather than directly.


## 11. Abstraction

**Abstraction** hides implementation details and exposes only the essential interface.

Python provides the `abc` module for formal abstract classes.

## Real-world analogy.
when you drive a car, you press:
    
     START
     
You don't need to understand:
   * Fuel injection
   *    ↓
   * Battery
   *    ↓
   * Starter motor
   *    ↓
   * Engine combustion
   *    ↓
   * Transmission


The complicated implementation is hidden.

Abstraction= show what is needed, hide how it works internally.
     

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("Car starts with a key or button.")

car = Car()
car.start()
````

### Output
```text
Car starts with a key or button.
```

An abstract method defines **what** a subclass must provide without defining the complete implementation.


## 12. Inheritance

Inheritance allows a child class to reuse and extend functionality from a parent class.

## Real-world analogy

A child can inherit characteristics from parents.

## Why?

To reuse existing code instead of writing it again.

## When?

When there is a genuine "is-a" relationship.

* Dog IS an Animal
* Car IS a Vehicle

### Basic structure
```text
Parent
  ↓
Child
```

````python
class Animal:

    def eat(self):
        print("Animal eats.")


class Dog(Animal):

    def bark(self):
        print("Dog barks.")


dog = Dog()

dog.eat()
dog.bark()
````

### Output
```text
Animal eats.
Dog barks.
```


## 13. Multiple Inheritance

A class inherits from more than one parent class.

## Real-world analogy

A person might have skills learned from different sources:

* Father → Driving
* Mother → Cooking
   *        ↓
*        Child

##  Why?

To combine behavior from multiple classes when the design genuinely requires it.

## When?

When one class logically needs functionality from multiple parent classes.

````python
class Father:

    def drive(self):
        print("Can drive.")


class Mother:

    def cook(self):
        print("Can cook.")


class Child(Father, Mother):
    pass


child = Child()

child.drive()
child.cook()
````

### Output
```text
can drive.
can cook.
```


## 14.Multilevel Inheritance

inheritance occcurs through multiple levels.

## Real-world analogy

* Grandparent
*    ↓
* Parent
*   ↓
* Child

## Why?

To represent a chain of specialization.

## When?

when each level is meaningful specialization of the previous level.


````python
class Grandparent:

    def family_name(self):
        print("Kumar")


class Parent(Grandparent):

    def job(self):
        print("Engineer")


class Child(Parent):

    def hobby(self):
        print("Coding")


child = Child()

child.family_name()
child.job()
child.hobby()
````

### Output
```text
kumar
Engineer
coding
```


## 15.Hierarchical Inheritance

multiple child classes inherit from one parent class.

## Real-world analogy
          Animal
          /    \
        Dog    Cat

## Why?
 To share common functionality among multiple related classes.

## When?
When several classes have the same common parent behavior.

````python
class Animal:

    def eat(self):
        print("Eating")


class Dog(Animal):

    def bark(self):
        print("Barking")


class Cat(Animal):

    def meow(self):
        print("Meowing")
````

## 16.Polymorphism

Polymorphism means "many forms."

The same interface or operation can produce different behaviour depending on the object.

## Real-world analogy

**Consider:**

Animal → speak()

**but:**
* Dog→ Woof
* Cat→ Meow
* Cow→ Moo

**Same operation:**
speak()

## Why?

it allows code to work with different objects types without needing separate logic for every type.

## When?
When different objects should respond to the same operation in their own way.

````python
class Dog:

    def speak(self):
        print("Woof")


class Cat:

    def speak(self):
        print("Meow")


dog = Dog()
cat = Cat()

dog.speak()
cat.speak()
````

### Output
```text
Woof
Meow
```

## 17. Method Overriding

Method overriding happens when a child class provides a different implementation of an inherited method.

## Real-world analogy

**Parent:**
"Animals make sounds."

**Dog:**
"My sound is Woof."

## Why?
To customize inherited behavior.

## When?
When the parent's implementation isn't suitable for a particualr child.

````python
class Animal:

    def speak(self):
        print("Animal makes a sound.")


class Dog(Animal):

    def speak(self):
        print("Dog says Woof.")


dog = Dog()
dog.speak()
````

## 18. Method overloading-Python Approach

Traditional method overloading means having multiple methods with the same name but different parameter lists.

python doesn't support traditional signature based overloading like java or c++.

instead, Python commonly uses:
* Default arguments
* args
* kwargs

## Real-world analogy
Think of an add() function.

you might want:
 add(10, 20)
 add(10, 20, 30)

 ## Why?

 to make one method flexible enough to handle different inputs.

 ## when?

 When the same operations should work with varying numbers or forms of inputs.

````python
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c


calculator = Calculator()

print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
````

### Output
```text
30
60
```

## 19. Composition
composition is a strong "has-a" relationship where one object contains another objects as part of its structure.

## Real-world analogy

A car has an engine

* Car
*  ↓
* Engine
The engine is the component of car.

## Why?

To build complex objects by combining smaller objects.

## When?

When one object is naturally made up of other objects.

````python
class Engine:

    def start(self):
        print("Engine started.")


class Car:

    def __init__(self):
        self.engine = Engine()

    def start(self):
        self.engine.start()
        print("Car started.")


car = Car()
car.start()
````

### Output
```text
Engine started.
Car started.
```
```

## 20.Aggregation

Aggregatiion is also a "has-a"relationship,but the contained objects can exist independently.

## Real-world analogy

A department has teacher

* Department
      ↓
*  Teachers
if the department closes, the teachers still exist.

## Why?

To represent relationship where objects are grouped together but don't have dependent lifetimes.

## When?

When one objects uses or groups other independently existing objects.

````python
class Teacher:

    def __init__(self, name):
        self.name = name


class Department:

    def __init__(self, teacher):
        self.teacher = teacher


teacher = Teacher("Mr. Kumar")

department = Department(teacher)

print(department.teacher.name)
````

### Output
```text
Mr.Kumar
```
```

## 21. Association

Association means two independent objects have relationship or interaction.

## Real-world analogy

A doctor treats a patient.

Doctor ↔️ Patient

The doctor doesn't own the patient.

The patient doesn't own the doctor.

They simply interact.

## Why?

To model Real-World relationships between objects.

## When?

Whenever two objects need to communicate or work together.

````python
class Doctor:

    def treat(self, patient):
        print("Doctor treats", patient.name)


class Patient:

    def __init__(self, name):
        self.name = name


doctor = Doctor()
patient = Patient("Aadharsh")

doctor.treat(patient)
````

### Output
```text
Doctor treats Aadharsh
```
```

## 23.Super()

Super() gives you access to the parent class implementation from a child class.

## Real-world analogy

Imagine you're a child and your parent already has a particular skill.

instead of rewriting that skill,you say:

"I'll use my parent's implementation."

## Why?
To reuse parent functionality.

## When?
Especially when extending a parent's constructor or method.

````python
class Person:

    def __init__(self, name):
        self.name = name


class Student(Person):

    def __init__(self, name, course):
        super().__init__(name)
        self.course = course


student = Student("Aadharsh", "AI & ML")

print(student.name)
print(student.course)
````

### Output
```text
Aadharsh
AI & ML
```
```

## 24.`__str__()`

`__str__()` defines the readable string representation of an object.

## Real-world analogy

if someone asks:
"Tell me about this employee."

You want:

Employee : Aadharsh, Role: AI&ML

rather than some technical memory address.

## Why?

To make objects readable when printed.

## when?

Whenever you want a clean, readable representation.


````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}, Age: {self.age}"


student = Student("Aadharsh", 21)

print(student)
````

### Output
```text
Aadharsh, Age: 21
```
```

## 25.`__repr__()`

`__repr__()` provides a more developer-oriented representation of an object.

## Real-world analogy

Think of two ways to describe a car.

**For a customer:**

Red toyota

**For a developer/debugger:**

car(brand='toyota', color='red', speed=80)

`__repr__()` is closer to the second idea.

## Why?

Useful for debugging and inspecting objects.

## When?

When developers need a detailed representation of objects.

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"Student(name='{self.name}', age={self.age})"


student = Student("Aadharsh", 21)

print(repr(student))
````

### Output
```text
Student(name='Aadharsh', age=21)
```
```

## 26. `__len__()`

`__len__()` defines what should happen when python's len() function is called on our object.

## Real-world analogy

suppose we create a team object.

We want:

len(team)

to return the number of team members.

## Why?

To make custom objects work naturally with Python's len() function.

## When?

when the object represents something that has a meaningful size or number of elements.

````python
class Team:

    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)


team = Team(["Aadharsh", "Rahul", "Priya"])

print(len(team))
````

### Output
```text
3
```
```

## 27.Dataclasses

A dataclass is a special Python feature designed to make classes that mainly store data easier to write.

## Real-world analogy

Supposee we're storing employee information:

Employee:

* Name
* ID
* Department
* Salary
we don't necessarily need a lot of custom behavior. We mainly need a clean data structure.

## Why?

it reduces boilerplate code.

Python can automatically generate things
such as:

* `__init__()`
* `__repr__()`
* `__eq__()`

## When?
When a class primarily represents structured data.


````python
from dataclasses import dataclass

@dataclass
class Student:
    name: str
    age: int
    marks: float


student = Student("Aadharsh", 21, 88.5)

print(student)
````

### Output
```text
Student(name='Aadharsh', age=21, marks=88.5)
```
```

## 28.Singleton pattern

Singleton is a design pattern that is intended to ensure that only one instance of a class exists.

## Real-world analogy

Imagine an application has a central configuration manager.

We want:

* Application
*   ↓
* One configuration manager

rather than creating multiple independent configuration managers.

## Why?

To provide a single shared instance where having multiple instances would be unnecessary or undesirable.

## When?

potential examples include:
* configuration management
* certain resource managers
* some application - wide services

however, singleton should be used carefully
because it introduces shared/global state.

````python
class Singleton:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)

        return cls._instance


obj1 = Singleton()
obj2 = Singleton()

print(obj1 is obj2)
````

### Output
```text
True
```
```
